# Validação do ambiente Spark + Postgres

Este notebook testa se o container do Spark consegue:
1. Criar uma SparkSession com o driver JDBC do Postgres
2. Escrever um DataFrame de teste no Postgres
3. Ler esse mesmo dado de volta

```python
# Importa a classe SparkSession da biblioteca PySpark.
# A SparkSession é o ponto de entrada para utilizar o Apache Spark.
from pyspark.sql import SparkSession

## Informa que o driver JDBC do PostgreSQL será baixado automaticamente
# pelo Spark através do Maven na primeira execução.

# Cria (ou reutiliza) uma SparkSession.
spark = (

    # Inicia o Builder da SparkSession.
    # O Builder é responsável por configurar a sessão antes de criá-la.
    SparkSession.builder

    # Define o nome da aplicação Spark.
    # Esse nome aparece na interface do Spark e nos logs.
    .appName("validacao-conexao")

    # Adiciona uma configuração à SparkSession.
    # Neste caso informa qual pacote Maven deverá ser carregado.
    .config(

        # Chave de configuração do Spark responsável
        # por carregar dependências externas.
        "spark.jars.packages",

        # Driver JDBC oficial do PostgreSQL.
        # Esse driver permite que o Spark converse com o banco.
        "org.postgresql:postgresql:42.7.3"
    )

    # Cria uma SparkSession caso ainda não exista.
    # Caso já exista uma sessão ativa, ela será reutilizada.
    .getOrCreate()
)

# Exibe uma mensagem informando que a SparkSession foi criada
# juntamente com a versão do Spark utilizada.
print("SparkSession criada com sucesso:", spark.version)
```

## Configuração do PostgreSQL

### Armazena em uma variável a URL de conexão JDBC utilizada pelo Spark.
### *jdbc:postgresql* informa que será utilizado o driver JDBC do PostgreSQL.
### *postgres* é o nome do servidor (host) do banco de dados.
### *5432* é a porta padrão do PostgreSQL.
### *etl_db* é o nome do banco de dados que será acessado.

```python
# Define a URL de conexão com o banco de dados.
jdbc_url = "jdbc:postgresql://postgres:5432/etl_db"

# Cria um dicionário (dict) contendo todas as propriedades
# necessárias para realizar a conexão com o banco.
jdbc_props = {

    # Define o usuário que será utilizado para autenticação no PostgreSQL.
    "user": "etl_user",

    # Define a senha do usuário informado acima.
    "password": "etl_pass",

    # Esse driver é responsável por permitir que o Spark converse com o banco.
    "driver": "org.postgresql.Driver",
}

# Criação de um DataFrame apenas para testar a conexão

# Cria um DataFrame do Spark a partir de dados existentes na memória.
df_teste = spark.createDataFrame(

    # Primeiro parâmetro:
    # Lista contendo três registros (linhas).
    # Cada tupla representa uma linha da futura tabela.
    [
        (1, "ok"),          # Primeira linha
        (2, "conexao"),     # Segunda linha
        (3, "validada")     # Terceira linha
    ],

    # Segundo parâmetro:
    # Define o nome das colunas do DataFrame.
    [
        "id",       # Primeira coluna
        "status"    # Segunda coluna
    ]
)

# Exibe o conteúdo do DataFrame no console.
# O método show() mostra os registros em formato de tabela.
df_teste.show()
```
